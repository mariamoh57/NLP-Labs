# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)  #why false ? to don't generate randomly

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

#this model only for french

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens  >> why ? to make the model faster
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text >> why ? So we can show it
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

## Task 2 — Encoder-only Model

Use the sentiment classifier on these three sentences:

```text
The movie was excellent.
The lecture was difficult to understand.
I enjoyed learning about transformers.
```

Print the predicted label and score for each sentence.

````python
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

sentences = [
    "The movie was excellent.",
    "The lecture was difficult to understand.",
    "I enjoyed learning about transformers."
]

for sentence in sentences:
    result = classifier(sentence)
    label = result[0]['label']
    score = result[0]['score']
    print(f"Sentence: '{sentence}'")
    print(f"Prediction: {label}, Score: {score:.4f}\n")
````

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

paragraph = """
In today's NLP lab, we explored Large Language Models (LLMs) and
the three main families of the Transformer architecture : The Decoder-only models like GPT for text generation,
Encoder-only models like BERT for sentiment classification, and Encoder-decoder models like T5 for tasks such as translation and summarization.
"""

prompt = f"Summarize the following paragraph:\n{paragraph}"

inputs = tokenizer(prompt, return_tensors="pt")
output = model.generate(**inputs, max_length=45)

summary = tokenizer.decode(output[0], skip_special_tokens=True)

print(f"\nOriginal Text: {paragraph.strip()}")
print(f"\nSummary: {summary}")

inputs1 = tokenizer(prompt, return_tensors="pt")
output1 = model.generate(**inputs1, max_length=60, min_length=20, no_repeat_ngram_size=2, num_beams=4)

summary1 = tokenizer.decode(output1[0], skip_special_tokens=True)

print(f"\nSummary: {summary1}")
````

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


**The answer**

it's a highly suitable for chatbots because its architecture is specifically designed for autoregressive text generation. It excels at predicting the next logical token based on all previous context, which perfectly matches the natural flow of a conversation. Unlike encoder models that need to process the entire input at once before generating, decoder models fluidly continue the text, allowing them to maintain context history and generate human-like, open-ended responses step by step.

### Question

**Hint:** Sampling introduces randomness when choosing the next token.


````python
generator = pipeline("text-generation", model="distilgpt2")
prompt = "Natural language processing is"

print("\nGeneration 1:")
out1 = generator(prompt, max_new_tokens=20, do_sample=True) # do_sample=True allows randomness
print(out1[0]["generated_text"])

print("\nGeneration 2:")
out2 = generator(prompt, max_new_tokens=20, do_sample=True)
print(out2[0]["generated_text"])

print("\nGeneration 3:")
out3 = generator(prompt, max_new_tokens=20, do_sample=True)
print(out3[0]["generated_text"])
````
